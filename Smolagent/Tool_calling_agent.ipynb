```python
from smolagents import CodeAgent , OpenAIModel, tool

@tool
def catering_service_tool(query: str) -> str:
    """
    This tool returns the highest-rated catering service in Gotham City.

    Args:
        query: A search term for finding catering services.
    """
    services = {
        "Gotham Catering Co. ": 4.9,
        "Wayne Manor Catering ": 4.8,
        "Gotham City Events ": 4.7
    }
    best_service = max(services, key = services.get)
    return best_service

model = OpenAIModel(
    model_id="local-model",
    api_base="http://127.0.0.1:1234/v1",  # Make sure /v1 is included
    api_key="lm-studio"
)

agent = CodeAgent(
    tools=[catering_service_tool],
    model=model
)
result = agent.run("can you give me the name of the highest-rated catering service in Gotham City?")

print(result)
```

```python
from smolagents import Tool, CodeAgent, OpenAIModel
class SuperheroPartyThemeTool(Tool):
    name = "superhero_party_theme_generator"
    description = """
    This tool suggest creative superhero-themed party ideas based on a category.
    It returns a unique party theme idea.
"""
    inputs = {
        "category":{
            "type": "string",
            "description": "The type of superhero party (e.g., 'classic heros', 'villain masquerade', 'futuristic Gotham')"
        }
    }
    output_type = "string"

    def forward(self, category: str):
        themes = {
            "classic heroes": "Justice League Gala: Guests come dressed as their favorite DC heroes with themed cocktails like 'The Kryptonite Punch'.",
            "villain masquerade": "Gotham Rogues' Ball: A mysterious masquerade where guests dress as classic Batman villains.",
            "futuristic Gotham": "Neo-Gotham Night: A cyberpunk-style party inspired by Batman Beyond, with neon decorations and futuristic gadgets."
        }

        return themes.get(category.lower(), "Themed party idea not found. Try 'classic heroes', 'villain masquerade', or 'futuristic Gotham'.")
    
party_theme_tool = SuperheroPartyThemeTool()
model = model = OpenAIModel(
    model_id="local-model",
    api_base="http://127.0.0.1:1234/v1", 
    api_key="lm-studio"
)
agent = CodeAgent(tools = [party_theme_tool], model = model)
result = agent.run("What would be a good superhero party idea for a 'villain masquerade' theme?")
print(result)
```

```python
import torch
from diffusers import StableDiffusionPipeline
from smolagents import CodeAgent, OpenAIModel, tool

# Load model
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
)
if torch.cuda.is_available():
    pipe = pipe.to("cuda")


@tool
def image_generator(description: str) -> str:
    """
    Generate an image from a text description using Stable Diffusion.

    Args:
        description: The text description of what image to generate.
    """
    # Force the local python interpreter block to run without interrupting your torch script
    image = pipe(description).images[0]
    filename = "generated_image.png"
    image.save(filename)
    return f"Image saved successfully as {filename}"


model = OpenAIModel(
    model_id="local-model", api_base="http://127.0.0.1:1234/v1", api_key="lm-studio"
)

# Initialize the agent normally. 
# By default, running a single tool inside the local runtime won't terminate 
# unless it triggers an external system process monitor.
agent = CodeAgent(
    tools=[image_generator],
    model=model,
)

agent.run(
    "Generate an image of a luxurious superhero-themed party at Wayne Manor"
)
```
